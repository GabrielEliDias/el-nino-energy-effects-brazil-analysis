```python
import urllib.request
import csv
import os  # <-- Importamos a biblioteca os
from pathlib import Path
```

```python

url = "https://psl.noaa.gov/data/correlation/oni.data"

pasta_destino = Path(r'C:\Users\joaof\Downloads\el-nino-energy-effects-brazil-analysis-main\el-nino-energy-effects-brazil-analysis-main\bronze')
output_csv = pasta_destino / "oni_dados.csv"

# Faz o download do arquivo
response = urllib.request.urlopen(url)
linhas = response.read().decode('utf-8').splitlines()


with open(output_csv, mode='w', newline='', encoding='utf-8') as file:
    writer = csv.writer(file)
    
    # Escreve o cabeçalho
    writer.writerow(["Ano", "Jan", "Fev", "Mar", "Abr", "Mai", "Jun", 
                     "Jul", "Ago", "Set", "Out", "Nov", "Dez"])
    
    # Processa e filtra as linhas
    for linha in linhas:
        valores = linha.split()
        
        if len(valores) == 13:
            writer.writerow(valores)

print(f"Concluído! Dados salvos em: '{output_csv}'")
```
